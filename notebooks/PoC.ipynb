# Peirce Continuum PoC — Demo Notebook

This notebook demonstrates the minimal Proof‑of‑Concept workflow: run a coarse sweep of the radial Dirac solver, plot the 1s binding energy vs. Z for the provided nucleus models, and inspect selected values. The notebook is intentionally compact so it can be run quickly as a first check.

## Quick instructions

1. Ensure the repository is cloned and you are in the repository root.
2. Install dependencies (once): `pip install -r requirements.txt`.
3. Run the cells below. The solver module is `src/dirac_radial_solver.py` and exposes `compute_E1s` and `sweep_Z`.

```python
# Imports
import os
import pandas as pd
import matplotlib.pyplot as plt

# Import the solver functions from src (the file must be present)
from src.dirac_radial_solver import compute_E1s, sweep_Z

plt.rcParams['figure.figsize'] = (8,5)
plt.rcParams['font.size'] = 10
```

## Run a coarse sweep (sphere model)

This runs a coarse sweep for Z in the chosen range. For a quick run, we use a reduced radial range and a coarse Z range. Adjust `Z_min`, `Z_max` and `r_max` for more accurate results.

```python
# Parameters for a quick demo run
Z_min = 100
Z_max = 200
R_fm = 5.0
r_max = 300.0
out_csv = 'data/E1s_sphere_R5fm_demo.csv'
os.makedirs('data', exist_ok=True)

# Run sweep (this may take some time depending on machine)
df_sphere = sweep_Z(Z_min, Z_max, model='sphere', R_fm=R_fm, out_csv=out_csv, r_max=r_max)
df_sphere.head()
```

## Plot E1s(Z)

Plot the computed 1s binding energies (in eV) and show the electron rest energy line for reference.

```python
df = df_sphere.copy()
plt.plot(df['Z'], df['E1s_eV'], label=f'sphere R={R_fm} fm')
plt.axhline(-510998.9461, color='k', linestyle='--', label='E = -m c^2 (eV)')
plt.xlabel('Z')
plt.ylabel('E1s (eV)')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

## Selected values

Show the computed energies for a few Z of interest (100, 137, 170, 180, 200).

```python
selected = [100, 137, 170, 180, 200]
sel_df = df[df['Z'].isin(selected)].set_index('Z')
sel_df[['E1s_au','E1s_eV','model','R_param']]
```

## Quick interpretation

- If the curve approaches the horizontal line at `E = -m c^2` and crosses it, that indicates the 1s state entering the negative continuum for that model and parameter set.
- For the homogeneous sphere model with R = 5 fm, the PoC runs typically show no crossing up to Z = 200 (model dependent). Use finer resolution and include Uehling/DF corrections for quantitative claims.
